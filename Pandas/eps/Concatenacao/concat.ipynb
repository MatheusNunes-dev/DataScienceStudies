```python
import pandas as pd 

df = pd.DataFrame({
    "cliente": [1,2,3,4,5],
    "nome": ["teo", "jose", "nah", "mah", "lah"],
})

df_02 = pd.DataFrame({
    "cliente": [6,7,8],
    "nome": ["kozato", "laura", "dan",],
    "idade":[32,29,31],
})

df_03 = pd.DataFrame({
    "idade": [32,34,19,54,33]
})
```

```python
# Concat serve para empilhar data frames
# Ignore index reseta o index
dfs = [df, df_02]
pd.concat(dfs, ignore_index=True)
```

```python
# Tambem pode cocatena um no lado do outro ( Praticamente um Join, mas sem foreign key, so usando a ordem da linhas)
# o Drop = True serve para que nao armazene o indexs antigos, se deixar o default False o index antes do sort vai ficar salvo em uma coluna
df_03 = df_03.sort_values(by=["idade"]).reset_index(drop=True)

new_df = pd.concat([df, df_03], axis=1)
new_df.head(5)
```

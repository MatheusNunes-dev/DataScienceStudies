```python
import pandas as pd

df = pd.read_csv("../../data/clientes.csv", sep=";")
df.head()
```

```python
# Para remover todos os NA ( NA = Null aqui)
# Tem criterios how

df.dropna()
df.dropna(how="all") # a linha inteira tem que ser na para remover
# O padrão é any, se tiver qualquer campo da linha for na vai remover a linha toda
```

```python
df_exemplo = pd.DataFrame({
    "nome" : ["Teo", None, "Nah", "Marcio"],
    "idade" : [None, None, 43, 52],
    "salario" : [3553, 4235, None, 5312]
})
```

```python
# Para remover Na especifico de alguma coluna usa-se subset com uma lista com o nome do campo(s)
df_exemplo.dropna(how="any", subset=["salario"])
```

```python
# Para preencher valores Na usa o fillna

df_exemplo.fillna(0) # Ou posso fillna na coluna 
df_exemplo["idade"].fillna(0)
```

```python
medidas = df_exemplo[["idade", "salario"]].mean()
type(medidas)
```

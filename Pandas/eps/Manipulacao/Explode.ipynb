```python
import pandas as pd

df = pd.read_csv("../../data/cartao.csv", sep= ";")
df
```

```python
df.columns = df.columns.str.strip()
df["dtTransacao"] = pd.to_datetime(df["dtTransacao"].str.strip())
df.rename(columns= {"vIVenda" : "vlVenda"})
```

```python
df.head()
```

```python
df["vlParcela"] = df["vIVenda"] / df["qtParcelas"]
```

```python
df["ordemParcela"] = df.apply(lambda row : [i for i in range(row["qtParcelas"])], axis=1) 
df_explode = df.explode(column = "ordemParcela")
```

```python
import pandas as pd

df = pd.read_csv("../../data/clientes.csv", sep=";")
df.head()
```

```python
# Para converter o tipo de uma coluna é só usar astype
df["qtdePontos"].astype(float)
df.head()
```

```python
df["DtCriacao"] = pd.to_datetime(df["DtCriacao"])
```

```python
# Exemplo de alterar o dado, usar um dicionario onde a key vai ser o valor antigo e o value vai ser o valor novo
# tem que usar atribuicao senao nao salva no df
df["DtCriacao"] = df["DtCriacao"].replace({"2025-11-13 12:03:56.114" : " 2024-02-01 09:00:00.000"})
```

```python
# Quando transfoma para date time da para acessar ano, mes, dia ...
df["DtCriacao"].dt.year
df["DtCriacao"].dt.day_of_week
```

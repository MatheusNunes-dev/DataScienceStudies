```python
import pandas as pd
# O segredo para passar o que voce quer emplihar é o set index
df = pd.DataFrame({
    "atletas" : ["Matheus", "João", "Luiz", "Arthur"],
    "atletismo" : [1, 2, 2, 1],
    "futebol" : [3, 2, 1, 4],
    "volei" : [0, 1, 1, 2]
})

df.head()
```

```python
# E interessante stack ( empilhar ), quando nao sabemos se tera mais colunas
# No caso poderia ter mais esportes
# Empilhando fica mais facil adcionar esportes novos, caso deixasse do outro jeito teria que crair uma nova coluna para cada novo esporte
df = df.set_index(["atletas"])
df = df.stack()
df.head()
# Isso retorna uma Series
```

```python
# Reset index faz ser um DataFrame
df = df.reset_index()
df.columns = ["Atleta", "Esporte", "Medalhas"]
df.head()
```

```python
# Unstack vai desimpilhar a o df
df = df.set_index(["Atleta", "Esporte"]).unstack().reset_index()
df.head()
```

```python
import pandas as pd

df = pd.read_csv("../../data/clientes.csv", sep=";")
df.head()
```

```python
exemplo = pd.DataFrame({
    "nome": ["teo", "lara", "nah", "bia", "mah", "lara", "mah", "mah"],
    "sobrenome" : ["calvo", "calvo", "ataide", "ataide", "silva", "silva", "silva", "silva"],
    "salario" : [1234, 5312, 5221, 2345, 1233, 6643, 4783, 6533]
})

exemplo.head()
```

```python
# Vai remover as linhas que são totalmente iguais ( Linha repetida ) - Por padrão so vai deixar a primeira 
exemplo.drop_duplicates()
exemplo.drop_duplicates(keep="last") # Se usar keep last vai remover as primeiras iguais e deixar somente a ultima
```

```python
# Se quiser analisar so colunas especificas com linhas duplicadas
exemplo.drop_duplicates(subset=["nome", "sobrenome"])
```

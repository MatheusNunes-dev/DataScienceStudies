```python
import pandas as pd
```

```python
# Criando uma Series

idades = [ 
    15, 16, 40, 35,
    27, 28, 20, 22,
    27, 61, 55, 12
]

series_idades = pd.Series(idades)
series_idades

```

```python
avg = series_idades.mean()
median = series_idades.median()
print(avg) 
print(median)
```

```python
#std = Desvio padrao - Mede o grau de dispersao dos dados em relacao a media
series_idades.describe()
```

```python
import pandas as pd 

df = pd.read_csv("../../data/clientes.csv", sep=";")
df.head()
```

```python
def get_last_id(id):
    return id.split("-")[-1]

df["idCliente"] = df["idCliente"].apply(func=get_last_id)
df.head()
```

```python
# Para conseguir transformar Salario em Float precisa primeiro deixar no formato de um float com uma funcao
exemplo = pd.DataFrame({
    "nome": ["teo", "lara", "nah", "bia", "mah", "lara", "mah", "mah"],
    "sobrenome" : ["calvo", "calvo", "ataide", "ataide", "silva", "silva", "silva", "silva"],
    "salario" : ["1 234,50"," 5312,00", "5221,13", "2345,41", "1 233,99", "6 643,53", "44 483", "6533,51"]
})
exemplo.head()
```

```python
def str_to_float(salario):
    salario = salario.replace(" ", "")
    salario = float(salario.replace(",", "."))
    return salario
    

exemplo["salario"] = exemplo["salario"].apply(func=str_to_float)
exemplo.dtypes
```

```python
def is_rich(x):
    return (x["salario"] > 5000 and x["sobrenome"] == "calvo")

exemplo["isRich"] = exemplo.apply(is_rich, axis= 1)
exemplo.head()
```

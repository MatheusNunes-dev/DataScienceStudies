```python
import pandas as pd

df_clientes = pd.read_csv("../../data/clientes.csv", sep=";")
df_clientes
```

## Amostras

```python
# .head exibi as primeiras linhas ( a quantidade de linhas que voce passar como argumento )
df_clientes.head(5)
```

```python
# .tail exibi as ultimas linhas do dataframe ( pode passar a quantidade como argumento )
df_clientes.tail(10)
```

```python
# .sample mostra uma amostra aleatoria do df
df_clientes.sample(5)
```

## Estrutura do DataFrame

```python
# .shape mostra a dimensao  ( numero de linhas, numero colunas ) do DataFrame
df_clientes.shape
```

```python
# .columns mostra o nome de todas as colunas do DataFrame
df_clientes.columns
```

```python
# .index retorna os index de um Dataframe ou de um Series
df_clientes.index
```

```python
# .info retorna um resumo do dataframe
df_clientes.info(memory_usage="deep")
```

```python
# .dtypes retorna uma series que tem o tipo de dado de cada coluna
df_clientes.dtypes
```

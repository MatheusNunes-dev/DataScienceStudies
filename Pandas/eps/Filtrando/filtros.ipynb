## Exemplo Simples

```python
brinquedo = pd.DataFrame(
    { 
        "nome": ["Matheus", "José", "João"],
        "idade": [20, 30, 25],
        "uf" : ["pr", "sp", "sc"]
    }
)
brinquedo
```

## Exemplo com transações

```python
filtro = df["QtdePontos"] >= 50
df[filtro]
```

```python
# O & é utilizado para And
filtro = (df["QtdePontos"] >= 50) & (df["QtdePontos"] <= 100)
df[filtro]
```

```python
# O | é utilizado para Or
filtro = (df["QtdePontos"] == 50) | (df["QtdePontos"] == 100)
df[filtro]
```

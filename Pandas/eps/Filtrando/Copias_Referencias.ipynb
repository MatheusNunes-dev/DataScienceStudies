# Cópias e Referências no Pandas

## Referência (mesmo objeto)

```python
df2 = df
```

* `df2` e `df` apontam para o **mesmo DataFrame**.
* Alterar `df2` também altera `df`.

---

## Cópia (novo objeto)

```python
df2 = df.copy()
```

* Cria um **novo DataFrame independente**.
* Alterações em `df2` **não afetam** `df`.

---

## Métodos do Pandas

```python
df2 = df.rename(...)
df2 = df.drop(...)
df2 = df.sort_values(...)
```

* Retornam um **novo DataFrame**.
* O DataFrame original **não é alterado**.
* Internamente, o Pandas pode compartilhar memória para otimização, mas isso é transparente para o usuário.

---

## Regra para decorar

| Código                | Resultado                   |
| --------------------- | --------------------------- |
| `df2 = df`            | Mesmo objeto (referência)   |
| `df2 = df.copy()`     | Novo DataFrame independente |
| `df.rename(...)`      | Retorna um novo DataFrame   |
| `df.drop(...)`        | Retorna um novo DataFrame   |
| `df.sort_values(...)` | Retorna um novo DataFrame   |

> **Dica:** Sempre que quiser modificar um DataFrame sem correr o risco de alterar o original, utilize `.copy()`.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([[-1, 3],[3, 2]], dtype=np.dtype(float))

b = np.array([7, 1], dtype=np.dtype(float))

print(f'Matrix A: {A}')
print(f'Shape Matrix A: {A.shape}')

print(f'Shape Array b: {b.shape}')
print(f'Array b: {b.shape}')

```

```python
# Solve Linear Algebra

x = np.linalg.solve(A, b)
print(f'X1, X2: {x}')
```

```python
# Finding the determinant
d = np.linalg.det(A)
print(f'Determinat: {d}')
```

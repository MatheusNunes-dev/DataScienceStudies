### Joing do SQL

```python
import pandas as pd 

transacoes = pd.read_csv("../../data/transacoes.csv", sep = ";")
transacoes = transacoes.rename(columns={"IdCliente": "idCliente"})
clientes = pd.read_csv("../../data/clientes.csv", sep=";")

transacoes.head()
```

```python
# Usamos right para falar a tabela do join, tem o how para colocar inner joio, left join ou right join or full join. 
# No on coloca a coluna que e a foreign key
# Como aqui a primary key e a forein key possuem o memso nome pode colocar so on
# Se a colunas estivessem com nomes diferentes precisa usar left_on e right_on
transacoes.merge(right=clientes, how="inner", on=["idCliente"] )
```

### Series
#### Uma estrutura de uma dimensão, com valores e índice. Uma coluna de um DataFrame é uma Series, mas ela também pode existir sozinha.

### DataFrame
#### Uma tabela com linhas e colunas. Cada coluna é uma Series, e as colunas compartilham o índice das linhas.

### loc
#### Seleciona pelo rótulo do índice e pelo nome da coluna. Em um intervalo, o último rótulo também é incluído.

### iloc (Integer Location)
#### Seleciona pela posição numérica da linha e da coluna: primeira = 0, segunda = 1… Em um intervalo, o último valor não é incluído.

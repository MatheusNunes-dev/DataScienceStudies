```python
import pandas as pd
idades = [ 
    15, 16, 40, 35,
    27, 28, 20, 22,
    27, 61, 55, 12
]

indexs = [ 
    'Matheus', 'João', 'Luiz', 'Andrea',
    'Gabriel', 'Eduardo', 'Arthur', 'Alexandre',
    'Thiago', 'Maria', 'Jose', 'Pedro'

]

series_idades = pd.Series(idades, index = indexs)
series_idades

```

```python
# Indice das Series funcionam da mesma forma que as chaves ( keys ) nos dicionarios 
series_idades[1]
```

```python
# Indices estão ligados a valores e nao a posicoes
sort_idades = series_idades.sort_values()
sort_idades
```

```python
# Quado se utiliza iloc fica como se fosse uma 'lista' ( nao pega a key e sim posicao)
# Sem iloc ( só loc ) está navegando nos indices e com iloc está navegando nas linhas
# Loc seleciona pelo Index
# Iloc seleciona pela posicao
sort_idades.iloc[0]
```

```python
sort_idades.iloc[:3]
```

```python
import pandas as pd

clientes = pd.read_csv("../../data/clientes.csv", sep=";")
clientes.head(3)
```

```python
# .sort_values ordena e retorna o series ordenada
clientes["qtdePontos"].sort_values()
```

```python
# Para ordenar e retornar um dataframe usa sort_values e by 
# Para ordernar de forma crescente usa ascending = True
# E ascending = False para decrescente
clientes.sort_values(by="qtdePontos", ascending=False).head(5)

# Posso fazer assim para pegar so uma coluna a partir de um Data Frame
# clientes.sort_values(by="qtdePontos", ascending=False).head(5)["IdCliente"]
```

```python
# Se quiser ordernar por mais de uma coluna 
clientes.sort_values(by=["qtdePontos", "flEmail"], ascending=[False, True]).head(5)
```

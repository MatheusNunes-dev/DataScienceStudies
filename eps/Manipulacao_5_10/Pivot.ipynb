```python
import pandas as pd

exemplo = pd.DataFrame({
    "nome": ["teo", "lara", "nah", "bia", "mah", "lara", "mah", "mah"],
    "sobrenome" : ["calvo", "calvo", "ataide", "ataide", "silva", "silva", "silva", "silva"],
    "idade" : [20, 25, 30, 35, 40, 45, 50, 55],
    "salario" : [1234, 5312, 5221, 2345, 1233, 6643, 4783, 6533]
})
```

```python
exemplo.pivot_table(values="salario", index="idade", aggfunc="mean")
exemplo.head()
```


Essas funções servem para **mudar o formato dos dados (reshape)**.

---

## 🔄 pivot → gira a tabela
- Reorganiza os dados
- Não aceita duplicados
- Não faz cálculos

👉 Coluna vira coluna

---

## 📊 pivot_table → gira + calcula
- Reorganiza os dados
- Aceita duplicados
- Faz cálculos (mean, sum, count...)

👉 Pivot com agregação

---

## 📦 stack → empilha
- Transforma colunas em linhas

👉 Colunas ↓ Linhas

---

## 📤 unstack → desempilha
- Transforma índices em colunas

👉 Linhas → Colunas

---

## 🧠 Resumo final

- pivot → gira tabela  
- pivot_table → gira + calcula  
- stack → empilha  
- unstack → desempilha
